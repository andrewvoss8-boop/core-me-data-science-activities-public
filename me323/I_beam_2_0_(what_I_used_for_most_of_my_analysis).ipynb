```python
"""
COMPLETE PHYSICS-INFORMED I-BEAM BAYESIAN OPTIMIZATION & VISUALIZATION
3D Search Space: (b, r, ΔH) with B fixed at 16mm

Usage:
1. Set EXPLORE_EXPLOIT_DIAL (0.0=greedy, 1.0=exploratory)
2. Toggle which recommendation sets to generate
3. Run to get recommendations and full visualization suite
"""

import pandas as pd
import numpy as np
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, ConstantKernel
from scipy.stats import norm
from scipy.optimize import minimize_scalar
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# ==========================================
# MASTER CONTROLS
# ==========================================
EXPLORE_EXPLOIT_DIAL = 0.05  # 0.0=very greedy, 1.0=very exploratory
GENERATE_MIXED = True    # Original 6-beam strategy (2+2+fantasy)
GENERATE_UCB = True      # Pure UCB (3 beams)
GENERATE_EI = True       # Pure EI (3 beams)
CSV_FILE = "I_beam_data.csv"
# ==========================================
# PHYSICS CONSTANTS
# ==========================================
TOTAL_HEIGHT = 25.0
B_FIXED = 16.0
MIN_WEB_THICKNESS = 0.8
MIN_FLANGE_WIDTH = 8.0  # ADD THIS LINE
MAX_WEB_RATIO = 2/3
MATERIAL_DENSITY = 1240
LENGTH_M = 0.2023
YIELD_STRENGTH = 76000000
BOUNDS_3D = {'b': (0.8, 10.67), 'r': (0.0, 4.0), 'delta_H': (-6.0, 4.0)}
PARAM_NAMES_3D = ['b_web (mm)', 'r_fillet (mm)', 'ΔH (mm)']

COL_MAP = {'H': 'H_web_height', 'B': 'B_flange_width', 'b': 'b_web_thick',
           'r': 'r_fillet', 'target': 'Str/w (N/g)'}

def get_acq_params(dial):
    """Generate acquisition parameters from explore/exploit dial"""
    xi_mod = 0.0001 + dial * 0.0199
    kappa_mod = 0.2 + dial * 1.8
    return {
        'EI': {'moderate': xi_mod, 'exploit': 0.0001, 'explore': 0.02},
        'UCB': {'moderate': kappa_mod, 'exploit': 0.2, 'explore': 2.0}
    }

ACQ_CONFIG = get_acq_params(EXPLORE_EXPLOIT_DIAL)

# ==========================================
# PHYSICS CALCULATIONS
# ==========================================
def calc_I(H, h, B, b):
    H_m, h_m, B_m, b_m = H/1000, h/1000, B/1000, b/1000
    I_web = (H_m**3 * b_m) / 12
    I_fl = (h_m**3 * B_m) / 12 + h_m * B_m * ((H_m + h_m) / 2)**2
    return I_web + 2*I_fl

def calc_mass(H, h, B, b):
    H_m, h_m, B_m, b_m = H/1000, h/1000, B/1000, b/1000
    return MATERIAL_DENSITY * LENGTH_M * (H_m*b_m + 2*h_m*B_m) * 1000

def calc_strength(H, h, B, b):
    return (4 * YIELD_STRENGTH * calc_I(H, h, B, b)) / (0.0125 * LENGTH_M)

def calc_str_w(H, B, b):
    h = (TOTAL_HEIGHT - H) / 2.0
    return calc_strength(H, h, B, b) / calc_mass(H, h, B, b)

def find_H_opt(b, B=B_FIXED):
    def obj(H):
        if H < 12.0 or H > 23.4: return 1e10
        h = (TOTAL_HEIGHT - H) / 2.0
        if h < 0 or h > 6.5: return 1e10
        return -calc_str_w(H, B, b)
    return minimize_scalar(obj, bounds=(12.0, 23.4), method='bounded').x

# ==========================================
# CONSTRAINTS & TRANSFORMS
# ==========================================
def check_3d(b, r, delta_H):
    H_phys = find_H_opt(b)
    H = H_phys + delta_H
    h = (TOTAL_HEIGHT - H) / 2.0
    if not (12.0 <= H <= 23.4 and 0 <= h <= 6.5): return False
    if not (MIN_WEB_THICKNESS <= b <= MAX_WEB_RATIO*B_FIXED): return False
    return 0 <= r <= (B_FIXED - b)/2.0

def enforce_3d(params):
    b, r, delta_H = params
    b = np.clip(b, BOUNDS_3D['b'][0], min(MAX_WEB_RATIO*B_FIXED, BOUNDS_3D['b'][1]))
    r = np.clip(r, 0, min((B_FIXED-b)/2.0, BOUNDS_3D['r'][1]))
    H_phys = find_H_opt(b)
    delta_H = np.clip(delta_H, BOUNDS_3D['delta_H'][0], BOUNDS_3D['delta_H'][1])
    H = np.clip(H_phys + delta_H, 12.0, 23.4)
    delta_H = H - H_phys
    h = (TOTAL_HEIGHT - H) / 2.0
    return np.array([b, r, delta_H]), H, h

def transform_4d_to_3d(X_4d):
    X_3d = []
    for i in range(len(X_4d)):
        b, r = X_4d[i, 2], X_4d[i, 3]
        H_phys = find_H_opt(b)
        delta_H = X_4d[i, 0] - H_phys
        X_3d.append([b, r, delta_H])
    return np.array(X_3d)

def normalize(X):
    X_n = np.copy(X).astype(float)
    for i, k in enumerate(['b', 'r', 'delta_H']):
        X_n[:, i] = (X[:, i] - BOUNDS_3D[k][0]) / (BOUNDS_3D[k][1] - BOUNDS_3D[k][0])
    return X_n

def denormalize(X_n):
    X = np.copy(X_n)
    for i, k in enumerate(['b', 'r', 'delta_H']):
        X[:, i] = X_n[:, i] * (BOUNDS_3D[k][1] - BOUNDS_3D[k][0]) + BOUNDS_3D[k][0]
    return X

# ==========================================
# GP TRAINING
# ==========================================
def train_gp(X_3d, y):
    X_norm = normalize(X_3d)
    y_log = np.log(y)
    y_mean, y_cent = np.mean(y_log), y_log - np.mean(y_log)

    kernel = ConstantKernel(1.0, (0.1, 10.0)) * Matern(
        length_scale=[0.5]*3, length_scale_bounds=(0.1, 3.0), nu=2.5)

    gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=25,
                                   alpha=1e-3, normalize_y=False)
    gp.fit(X_norm, y_cent)
    return gp, X_norm, y_cent, y_mean

# ==========================================
# ACQUISITION FUNCTIONS
# ==========================================
def ei_acq(gp, X_norm, y_best, xi):
    mu, sigma = gp.predict(X_norm, return_std=True)
    sigma = np.maximum(sigma, 1e-9)
    imp = mu - y_best - xi
    Z = imp / sigma
    return imp * norm.cdf(Z) + sigma * norm.pdf(Z), mu, sigma

def ucb_acq(gp, X_norm, kappa):
    mu, sigma = gp.predict(X_norm, return_std=True)
    return mu + kappa * np.maximum(sigma, 1e-9), mu, sigma

def optimize_acq(gp, acq_fn, acq_params, n_cand=4000):
    valid_cand, valid_norm = [], []
    attempts = 0
    while len(valid_cand) < n_cand and attempts < 5000:
        attempts += 1
        b = np.random.uniform(*BOUNDS_3D['b'])
        r = np.random.uniform(0, min((B_FIXED-b)/2.0, BOUNDS_3D['r'][1]))
        dH = np.random.uniform(*BOUNDS_3D['delta_H'])
        if check_3d(b, r, dH):
            cand = np.array([b, r, dH])
            valid_cand.append(cand)
            valid_norm.append(normalize(cand.reshape(1, -1))[0])

    if not valid_cand: raise ValueError("No valid candidates!")
    valid_norm = np.array(valid_norm)
    acq_vals, mu_vals, sig_vals = acq_fn(gp, valid_norm, **acq_params)
    idx = np.argmax(acq_vals)
    return valid_cand[idx], acq_vals[idx], mu_vals[idx], sig_vals[idx]

# ==========================================
# RECOMMENDATION GENERATORS
# ==========================================
def gen_mixed(gp, y_cent, y_mean, X_train_3d, y_train):
    np.random.seed(100)  # Same seed as mixed beam 1

    print(f"\n{'='*70}\nMIXED: 2 EI + 2 UCB moderate, then 2 fantasy\n{'='*70}")
    recs, y_best = [], y_cent.max()

    # Beam 1: EI moderate
    xi = ACQ_CONFIG['EI']['moderate']
    x, _, mu, sig = optimize_acq(gp, ei_acq, {'y_best': y_best, 'xi': xi})
    x, H, h = enforce_3d(x)
    recs.append({'Beam': 1, 'Method': 'EI', 'Param': f'ξ={xi:.4f}', 'Type': 'Mod',
                 'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
                 'Pred': np.exp(mu + y_mean), 'Var': sig})

    # Beam 2: UCB moderate
    kappa = ACQ_CONFIG['UCB']['moderate']
    x, _, mu, sig = optimize_acq(gp, ucb_acq, {'kappa': kappa})
    x, H, h = enforce_3d(x)
    recs.append({'Beam': 2, 'Method': 'UCB', 'Param': f'κ={kappa:.4f}', 'Type': 'Mod',
                 'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
                 'Pred': np.exp(mu + y_mean), 'Var': sig})

    # Fantasy GP
    X_fant = np.vstack([X_train_3d, [recs[0]['b'], recs[0]['r'], recs[0]['dH']],
                        [recs[1]['b'], recs[1]['r'], recs[1]['dH']]])
    y_fant = np.append(y_train, [recs[0]['Pred'], recs[1]['Pred']])
    gp_f, _, y_f_cent, y_f_mean = train_gp(X_fant, y_fant)
    y_best_f = y_f_cent.max()

    # Beams 3-6
    for i, (meth, typ) in enumerate([('EI','exploit'),('EI','explore'),
                                      ('UCB','exploit'),('UCB','explore')], 3):
        if meth == 'EI':
            xi = ACQ_CONFIG['EI'][typ]
            x, _, mu, sig = optimize_acq(gp_f, ei_acq, {'y_best': y_best_f, 'xi': xi})
            pstr = f'ξ={xi:.4f}'
        else:
            kappa = ACQ_CONFIG['UCB'][typ]
            x, _, mu, sig = optimize_acq(gp_f, ucb_acq, {'kappa': kappa})
            pstr = f'κ={kappa:.4f}'
        x, H, h = enforce_3d(x)
        recs.append({'Beam': i, 'Method': meth, 'Param': pstr, 'Type': typ.title(),
                     'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
                     'Pred': np.exp(mu + y_f_mean), 'Var': sig})

    return pd.DataFrame(recs)

def gen_ucb(gp, y_mean, X_train_3d, y_train, y_cent_orig):
    np.random.seed(100)
    print(f"\n{'='*70}\nPURE UCB: 3 beams (WITH FANTASY)\n{'='*70}")
    recs, kappa = [], ACQ_CONFIG['UCB']['moderate']

    # Use original GP and data for tracking
    current_gp = gp
    current_X = X_train_3d.copy()
    current_y = y_train.copy()
    current_y_mean = y_mean

    for i in range(3):
        x, _, mu, sig = optimize_acq(current_gp, ucb_acq, {'kappa': kappa}, n_cand=4000)
        x, H, h = enforce_3d(x)
        pred = np.exp(mu + current_y_mean)

        recs.append({'Beam': i+1, 'Method': 'UCB', 'Param': f'κ={kappa:.4f}', 'Type': 'Mod',
                     'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
                     'Pred': pred, 'Var': sig})

        # After each beam, update fantasy GP (except after last beam)
        if i < 2:
            current_X = np.vstack([current_X, [x[0], x[1], x[2]]])
            current_y = np.append(current_y, pred)
            current_gp, _, _, current_y_mean = train_gp(current_X, current_y)

    return pd.DataFrame(recs)

def gen_ei(gp, y_cent, y_mean, X_train_3d, y_train):
    np.random.seed(100)
    print(f"\n{'='*70}\nPURE EI: 3 beams (WITH FANTASY)\n{'='*70}")
    recs, xi = [], ACQ_CONFIG['EI']['moderate']

    # Use original GP and data for tracking
    current_gp = gp
    current_X = X_train_3d.copy()
    current_y = y_train.copy()
    current_y_mean = y_mean
    current_y_cent = y_cent.copy()

    for i in range(3):
        y_best = current_y_cent.max()
        x, _, mu, sig = optimize_acq(current_gp, ei_acq, {'y_best': y_best, 'xi': xi}, n_cand=4000)
        x, H, h = enforce_3d(x)
        pred = np.exp(mu + current_y_mean)

        recs.append({'Beam': i+1, 'Method': 'EI', 'Param': f'ξ={xi:.4f}', 'Type': 'Mod',
                     'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
                     'Pred': pred, 'Var': sig})

        # After each beam, update fantasy GP (except after last beam)
        if i < 2:
            current_X = np.vstack([current_X, [x[0], x[1], x[2]]])
            current_y = np.append(current_y, pred)
            current_gp, _, current_y_cent, current_y_mean = train_gp(current_X, current_y)

    return pd.DataFrame(recs)

# ==========================================
# PLOTTING
# ==========================================
def get_point_sizes(X_norm, slice_point_norm, max_distance=0.5):
    """
    Compute marker sizes based on distance from slice using buckets.
    Points are categorized into proximity tiers with distinct sizes.
    """
    distances = np.linalg.norm(X_norm - slice_point_norm, axis=1)

    sizes = np.zeros_like(distances)
    colors = np.zeros_like(distances)

    # Assign sizes based on buckets
    bucket1 = distances < 0.15
    bucket2 = (distances >= 0.15) & (distances < 0.30)
    bucket3 = (distances >= 0.30) & (distances < 0.50)
    bucket4 = (distances >= 0.50) & (distances < 0.75)
    bucket5 = distances >= 0.75

    sizes[bucket1] = 250
    sizes[bucket2] = 180
    sizes[bucket3] = 120
    sizes[bucket4] = 70
    sizes[bucket5] = 30

    colors[bucket1] = 0.9
    colors[bucket2] = 0.7
    colors[bucket3] = 0.5
    colors[bucket4] = 0.3
    colors[bucket5] = 0.1

    return sizes, distances, colors

def plot_1d_marginal(gp, X_norm, y, y_mean, rec_list, n_samples=200):
    """
    Plot 1D marginal effects by averaging over random samples of other dimensions.
    This shows the true average effect of each parameter across the design space.
    """
    print("\nPlotting 1D marginal effects (averaged across all slices)...")

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    fig.suptitle(f'I-Beam Design (B={B_FIXED}mm): Marginal Effects (Averaged Across Design Space)',
                 fontsize=14, weight='bold')

    for dim in range(3):
        ax = axes[dim]
        x_vals_norm = np.linspace(0, 1, 100)

        # For each value of this dimension, sample random values for the other dimensions
        mu_samples_all = []
        sig_samples_all = []

        for x_val in x_vals_norm:
            # Generate random samples for the other dimensions
            X_samples = np.random.rand(n_samples, 3)
            X_samples[:, dim] = x_val  # Fix this dimension

            # Filter to only valid designs
            valid_samples = []
            for sample in X_samples:
                b, r, dH = denormalize(sample.reshape(1, -1))[0]
                if check_3d(b, r, dH):
                    valid_samples.append(sample)

            if len(valid_samples) > 0:
                valid_samples = np.array(valid_samples)
                mu, sig = gp.predict(valid_samples, return_std=True)
                mu_samples_all.append(mu)
                sig_samples_all.append(sig)
            else:
                # If no valid samples, use NaN
                mu_samples_all.append(np.array([np.nan]))
                sig_samples_all.append(np.array([np.nan]))

        # Compute mean and std of means/uncertainties across samples
        mu_avg = np.array([np.nanmean(m) for m in mu_samples_all])
        mu_std = np.array([np.nanstd(m) for m in mu_samples_all])
        sig_avg = np.array([np.nanmean(s) for s in sig_samples_all])

# Convert back to real space
        mu_r = np.exp(mu_avg + y_mean)
        epistemic_std = np.exp(mu_avg + y_mean) * sig_avg
        mu_variability = np.exp(mu_avg + y_mean) * mu_std

        # Properly denormalize x-axis values for current dimension
        keys = list(BOUNDS_3D.keys())
        x_vals = x_vals_norm * (BOUNDS_3D[keys[dim]][1] - BOUNDS_3D[keys[dim]][0]) + BOUNDS_3D[keys[dim]][0]

        # Plot mean with epistemic uncertainty
        ax.plot(x_vals, mu_r, 'b-', lw=2, label='Marginal mean')



        ax.fill_between(x_vals, mu_r - 2*epistemic_std, mu_r + 2*epistemic_std,
                        alpha=0.3, color='blue', label='±2σ epistemic (avg)')

        # Plot variability of mean across design space
        ax.fill_between(x_vals, mu_r - mu_variability, mu_r + mu_variability,
                        alpha=0.15, color='purple', label='Mean variability')

        # Plot data
        X_den = denormalize(X_norm)
        ax.scatter(X_den[:, dim], y, c='red', s=50, alpha=0.6, ec='black', lw=1,
                  label='Data', zorder=5)

        # Plot recommendations
        for rec, clr, mrk, lbl in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D'],
                                       ['Mixed','UCB','EI']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                ax.scatter(rec_3d[:, dim], rec['Pred'], c=clr, s=150, marker=mrk,
                          ec='black', lw=1.5, label=lbl, zorder=10)

        if dim == 2:
            ax.axvline(0, color='green', ls='--', alpha=0.5, lw=2, label='Physics opt')

        ax.set_xlabel(PARAM_NAMES_3D[dim], fontsize=11)
        ax.set_ylabel('Str/w (N/g)', fontsize=11)
        ax.set_title(PARAM_NAMES_3D[dim], fontsize=12, weight='bold')
        ax.legend(fontsize=8, loc='best')
        ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.savefig('ibeam_1d_marginal_B16mm.png', dpi=150, bbox_inches='tight')
    plt.show()

def plot_2d_marginal(gp, X_norm, y, y_mean, rec_list, n_samples=100):
    """
    Plot 2D marginal effects by averaging over random samples of the third dimension.
    """
    print("\nPlotting 2D marginal contours (averaged across third dimension)...")

    combos = [(0,1,'b_web','r_fillet',2), (0,2,'b_web','ΔH',1), (1,2,'r_fillet','ΔH',0)]

    for d1, d2, n1, n2, d_fixed in combos:
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))
        fig.suptitle(f'I-Beam (B={B_FIXED}mm): Marginal {n1} vs {n2}\n' +
                     f'(Averaged over {PARAM_NAMES_3D[d_fixed]})',
                     fontsize=14, weight='bold')

        res = 40  # Lower resolution due to sampling
        grid_x = np.linspace(0, 1, res)
        grid_y = np.linspace(0, 1, res)

        mu_grid = np.zeros((res, res))
        sig_grid = np.zeros((res, res))

        print(f"  Processing {n1} vs {n2}...")
        for i, x_val in enumerate(grid_x):
            for j, y_val in enumerate(grid_y):
                # Sample random values for the fixed dimension
                X_samples = np.random.rand(n_samples, 3)
                X_samples[:, d1] = x_val
                X_samples[:, d2] = y_val

                # Filter valid samples
                valid_samples = []
                for sample in X_samples:
                    b, r, dH = denormalize(sample.reshape(1, -1))[0]
                    if check_3d(b, r, dH):
                        valid_samples.append(sample)

                if len(valid_samples) > 0:
                    valid_samples = np.array(valid_samples)
                    mu, sig = gp.predict(valid_samples, return_std=True)
                    mu_grid[j, i] = np.mean(mu)
                    sig_grid[j, i] = np.mean(sig)
                else:
                    mu_grid[j, i] = np.nan
                    sig_grid[j, i] = np.nan

        # Convert to real space
        mu_r = np.exp(mu_grid + y_mean)
        sig_r = sig_grid

        keys = list(BOUNDS_3D.keys())
        x_vals = grid_x * (BOUNDS_3D[keys[d1]][1] - BOUNDS_3D[keys[d1]][0]) + BOUNDS_3D[keys[d1]][0]
        y_vals = grid_y * (BOUNDS_3D[keys[d2]][1] - BOUNDS_3D[keys[d2]][0]) + BOUNDS_3D[keys[d2]][0]
        ext = [x_vals[0], x_vals[-1], y_vals[0], y_vals[-1]]

        # Mean
        im1 = axes[0].imshow(mu_r, origin='lower', extent=ext, aspect='auto', cmap='viridis')
        axes[0].contour(mu_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[0].set_xlabel(n1+' (mm)', fontsize=12)
        axes[0].set_ylabel(n2+' (mm)', fontsize=12)
        axes[0].set_title('Marginal Mean', fontsize=13, weight='bold')

        X_den = denormalize(X_norm)
        axes[0].scatter(X_den[:, d1], X_den[:, d2], c='red', s=80, ec='white', lw=2,
                       label='Data', zorder=10)

        for rec, clr, mrk, lbl in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D'],
                                       ['Mixed','UCB','EI']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                axes[0].scatter(rec_3d[:, d1], rec_3d[:, d2], c=clr, s=150, marker=mrk,
                              ec='black', lw=1.5, label=lbl, zorder=11)

        if d2 == 2: axes[0].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[0].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        axes[0].legend(fontsize=9)
        plt.colorbar(im1, ax=axes[0], label='Str/w (N/g)')

        # Uncertainty
        im2 = axes[1].imshow(sig_r, origin='lower', extent=ext, aspect='auto', cmap='hot')
        axes[1].contour(sig_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[1].set_xlabel(n1+' (mm)', fontsize=12)
        axes[1].set_ylabel(n2+' (mm)', fontsize=12)
        axes[1].set_title('Marginal Uncertainty (σ)', fontsize=13, weight='bold')
        axes[1].scatter(X_den[:, d1], X_den[:, d2], c='cyan', s=80, ec='white', lw=2, zorder=10)

        for rec, clr, mrk in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                axes[1].scatter(rec_3d[:, d1], rec_3d[:, d2], c=clr, s=150, marker=mrk,
                              ec='black', lw=1.5, zorder=11)

        if d2 == 2: axes[1].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[1].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        plt.colorbar(im2, ax=axes[1], label='Std Dev (σ)')

        plt.tight_layout()
        plt.savefig(f'ibeam_2d_marginal_{n1}_{n2}_B16mm.png', dpi=150, bbox_inches='tight')
        plt.show()

def plot_1d_global(gp, X_norm, y, y_mean, rec_list):
    """Plot 1D slices at mean values (global view)"""
    print("\nPlotting 1D global slices...")
    slice_pt = np.mean(X_norm, axis=0)
    slice_real = denormalize(slice_pt.reshape(1, -1))[0]

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    fig.suptitle(f'I-Beam Design (B={B_FIXED}mm): Global 1D Slices\n' +
                 f'(Fixed at: b={slice_real[0]:.2f}mm, r={slice_real[1]:.2f}mm, ΔH={slice_real[2]:.2f}mm)',
                 fontsize=14, weight='bold')

    for dim in range(3):
        ax = axes[dim]
        X_test = np.tile(slice_pt, (100, 1))
        X_test[:, dim] = np.linspace(0, 1, 100)

        mu, sig = gp.predict(X_test, return_std=True)
        mu_r = np.exp(mu + y_mean)
        epistemic_std = np.exp(mu + y_mean) * sig

        x_vals = denormalize(X_test)[:, dim]

        ax.plot(x_vals, mu_r, 'b-', lw=2, label='Mean prediction')
        ax.fill_between(x_vals, mu_r - 2*epistemic_std, mu_r + 2*epistemic_std,
                        alpha=0.3, color='blue', label='±2σ epistemic')

        X_den = denormalize(X_norm)
        ax.scatter(X_den[:, dim], y, c='red', s=50, alpha=0.6, ec='black', lw=1,
                  label='Data', zorder=5)

        for rec, clr, mrk, lbl in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D'],
                                       ['Mixed','UCB','EI']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                ax.scatter(rec_3d[:, dim], rec['Pred'], c=clr, s=150, marker=mrk,
                          ec='black', lw=1.5, label=lbl, zorder=10)

        if dim == 2:
            ax.axvline(0, color='green', ls='--', alpha=0.5, lw=2, label='Physics opt')

        ax.set_xlabel(PARAM_NAMES_3D[dim], fontsize=11)
        ax.set_ylabel('Str/w (N/g)', fontsize=11)
        ax.set_title(PARAM_NAMES_3D[dim], fontsize=12, weight='bold')
        ax.legend(fontsize=8)
        ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.savefig('ibeam_1d_global_B16mm.png', dpi=150, bbox_inches='tight')
    plt.show()


def plot_2d_global(gp, X_norm, y, y_mean, rec_list):
    """Plot 2D contours at mean values (global view)"""
    print("\nPlotting 2D global contours...")
    slice_pt = np.mean(X_norm, axis=0)
    slice_real = denormalize(slice_pt.reshape(1, -1))[0]

    combos = [(0,1,'b_web','r_fillet',2), (0,2,'b_web','ΔH',1), (1,2,'r_fillet','ΔH',0)]

    for d1, d2, n1, n2, d_fixed in combos:
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))
        fig.suptitle(f'I-Beam (B={B_FIXED}mm): {n1} vs {n2}\n' +
                     f'(Fixed: {PARAM_NAMES_3D[d_fixed]}={slice_real[d_fixed]:.2f}mm)',
                     fontsize=14, weight='bold')

        grid_x, grid_y = np.meshgrid(np.linspace(0, 1, 50), np.linspace(0, 1, 50))
        X_test = np.tile(slice_pt, (2500, 1))
        X_test[:, d1], X_test[:, d2] = grid_x.ravel(), grid_y.ravel()

        mu, sig = gp.predict(X_test, return_std=True)
        mu_r = np.exp(mu + y_mean).reshape(50, 50)
        sig_r = sig.reshape(50, 50)

        keys = list(BOUNDS_3D.keys())
        x_v = np.linspace(0, 1, 50) * (BOUNDS_3D[keys[d1]][1] - BOUNDS_3D[keys[d1]][0]) + BOUNDS_3D[keys[d1]][0]
        y_v = np.linspace(0, 1, 50) * (BOUNDS_3D[keys[d2]][1] - BOUNDS_3D[keys[d2]][0]) + BOUNDS_3D[keys[d2]][0]
        ext = [x_v[0], x_v[-1], y_v[0], y_v[-1]]

        # Mean
        im1 = axes[0].imshow(mu_r, origin='lower', extent=ext, aspect='auto', cmap='viridis')
        axes[0].contour(mu_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[0].set_xlabel(n1+' (mm)', fontsize=12)
        axes[0].set_ylabel(n2+' (mm)', fontsize=12)
        axes[0].set_title('Mean Prediction', fontsize=13, weight='bold')

        X_den = denormalize(X_norm)
        axes[0].scatter(X_den[:, d1], X_den[:, d2], c='red', s=80, ec='white', lw=2,
                       label='Data', zorder=10)

        for rec, clr, mrk, lbl in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D'],
                                       ['Mixed','UCB','EI']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                axes[0].scatter(rec_3d[:, d1], rec_3d[:, d2], c=clr, s=150, marker=mrk,
                              ec='black', lw=1.5, label=lbl, zorder=11)

        if d2 == 2: axes[0].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[0].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        axes[0].legend(fontsize=9)
        plt.colorbar(im1, ax=axes[0], label='Str/w (N/g)')

        # Uncertainty
        im2 = axes[1].imshow(sig_r, origin='lower', extent=ext, aspect='auto', cmap='hot')
        axes[1].contour(sig_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[1].set_xlabel(n1+' (mm)', fontsize=12)
        axes[1].set_ylabel(n2+' (mm)', fontsize=12)
        axes[1].set_title('Uncertainty (σ)', fontsize=13, weight='bold')
        axes[1].scatter(X_den[:, d1], X_den[:, d2], c='cyan', s=80, ec='white', lw=2, zorder=10)

        for rec, clr, mrk in zip(rec_list, ['lime','cyan','yellow'], ['*','s','D']):
            if rec is not None:
                rec_3d = rec[['b','r','dH']].values
                axes[1].scatter(rec_3d[:, d1], rec_3d[:, d2], c=clr, s=150, marker=mrk,
                              ec='black', lw=1.5, zorder=11)

        if d2 == 2: axes[1].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[1].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        plt.colorbar(im2, ax=axes[1], label='Std Dev (σ)')

        plt.tight_layout()
        plt.savefig(f'ibeam_2d_global_{n1}_{n2}_B16mm.png', dpi=150, bbox_inches='tight')
        plt.show()

def plot_slice_1d(gp, X_norm, y, y_mean, slice_point_norm, slice_name, slice_idx):
    """Plot 1D slices at a specific point"""
    slice_real = denormalize(slice_point_norm.reshape(1, -1))[0]
    point_sizes, point_distances, point_colors = get_point_sizes(X_norm, slice_point_norm)
    on_slice = point_distances < 0.1

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    fig.suptitle(f'Slice {slice_idx}: {slice_name}\n' +
                 f'(b={slice_real[0]:.2f}mm, r={slice_real[1]:.2f}mm, ΔH={slice_real[2]:.2f}mm)',
                 fontsize=14, weight='bold')

    print(f"  Slice {slice_idx} - Point distribution:")
    print(f"    Very close (<0.15): {np.sum(point_distances < 0.15)}")
    print(f"    Close (0.15-0.30): {np.sum((point_distances >= 0.15) & (point_distances < 0.30))}")
    print(f"    Moderate (0.30-0.50): {np.sum((point_distances >= 0.30) & (point_distances < 0.50))}")
    print(f"    Far (>0.50): {np.sum(point_distances >= 0.50)}")

    for dim in range(3):
        ax = axes[dim]
        X_test = np.tile(slice_point_norm, (100, 1))
        X_test[:, dim] = np.linspace(0, 1, 100)

        mu, sig = gp.predict(X_test, return_std=True)
        mu_r = np.exp(mu + y_mean)
        epistemic_std = np.exp(mu + y_mean) * sig

        x_vals = denormalize(X_test)[:, dim]

        ax.plot(x_vals, mu_r, 'b-', lw=2, label='Mean')
        ax.fill_between(x_vals, mu_r - 2*epistemic_std, mu_r + 2*epistemic_std,
                        alpha=0.3, color='blue', label='±2σ epistemic')

        X_den = denormalize(X_norm)
        ax.scatter(X_den[:, dim], y, s=point_sizes, c=point_colors, cmap='Reds',
                  vmin=0, vmax=1, alpha=0.6, ec='black', lw=0.5, zorder=5,
                  label='Data (sized by distance)')

        if np.any(on_slice):
            ax.scatter(X_den[on_slice, dim], y[on_slice], s=400, marker='*',
                      c='yellow', ec='black', lw=2, zorder=10, label='On slice')

        if dim == 2:
            ax.axvline(0, color='green', ls='--', alpha=0.5, lw=2, label='Physics opt')

        ax.set_xlabel(PARAM_NAMES_3D[dim], fontsize=11)
        ax.set_ylabel('Str/w (N/g)', fontsize=11)
        ax.set_title(PARAM_NAMES_3D[dim], fontsize=12, weight='bold')
        ax.legend(fontsize=8)
        ax.grid(alpha=0.3)

    plt.tight_layout()
    clean_name = slice_name.replace('#', 'num').replace('(', '').replace(')', '').replace('=', '').replace(':', '').replace('/', '_').replace(' ', '_')
    plt.savefig(f'ibeam_slice_{slice_idx:02d}_{clean_name}_1d_B16mm.png', dpi=150, bbox_inches='tight')
    plt.show()

def plot_slice_2d(gp, X_norm, y, y_mean, slice_point_norm, slice_name, slice_idx):
    """Plot 2D contours at a specific point"""
    slice_real = denormalize(slice_point_norm.reshape(1, -1))[0]
    point_sizes, point_distances, point_colors = get_point_sizes(X_norm, slice_point_norm)
    on_slice = point_distances < 0.1

    combos = [(0,1,'b_web','r_fillet',2), (0,2,'b_web','ΔH',1), (1,2,'r_fillet','ΔH',0)]

    for d1, d2, n1, n2, d_fixed in combos:
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))
        fig.suptitle(f'Slice {slice_idx}: {slice_name} - {n1} vs {n2}\n' +
                     f'(Fixed: {PARAM_NAMES_3D[d_fixed]}={slice_real[d_fixed]:.2f}mm)',
                     fontsize=14, weight='bold')

        grid_x, grid_y = np.meshgrid(np.linspace(0, 1, 50), np.linspace(0, 1, 50))
        X_test = np.tile(slice_point_norm, (2500, 1))
        X_test[:, d1], X_test[:, d2] = grid_x.ravel(), grid_y.ravel()

        mu, sig = gp.predict(X_test, return_std=True)
        mu_r = np.exp(mu + y_mean).reshape(50, 50)
        sig_r = sig.reshape(50, 50)

        keys = list(BOUNDS_3D.keys())
        x_v = np.linspace(0, 1, 50) * (BOUNDS_3D[keys[d1]][1] - BOUNDS_3D[keys[d1]][0]) + BOUNDS_3D[keys[d1]][0]
        y_v = np.linspace(0, 1, 50) * (BOUNDS_3D[keys[d2]][1] - BOUNDS_3D[keys[d2]][0]) + BOUNDS_3D[keys[d2]][0]
        ext = [x_v[0], x_v[-1], y_v[0], y_v[-1]]

        # Mean
        im1 = axes[0].imshow(mu_r, origin='lower', extent=ext, aspect='auto', cmap='viridis')
        axes[0].contour(mu_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[0].set_xlabel(n1+' (mm)', fontsize=12)
        axes[0].set_ylabel(n2+' (mm)', fontsize=12)
        axes[0].set_title('Mean Prediction', fontsize=12, weight='bold')

        X_den = denormalize(X_norm)
        axes[0].scatter(X_den[:, d1], X_den[:, d2], s=point_sizes, c=point_colors,
                       cmap='Reds', vmin=0, vmax=1, alpha=0.6, ec='white', lw=1, zorder=5)

        if np.any(on_slice):
            axes[0].scatter(X_den[on_slice, d1], X_den[on_slice, d2], s=400, marker='*',
                           c='yellow', ec='black', lw=2, zorder=10, label='On slice')
            axes[0].legend()

        if d2 == 2: axes[0].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[0].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        plt.colorbar(im1, ax=axes[0], label='Str/w (N/g)')

        # Uncertainty
        im2 = axes[1].imshow(sig_r, origin='lower', extent=ext, aspect='auto', cmap='hot')
        axes[1].contour(sig_r, origin='lower', extent=ext, colors='white', alpha=0.4, levels=10)
        axes[1].set_xlabel(n1+' (mm)', fontsize=12)
        axes[1].set_ylabel(n2+' (mm)', fontsize=12)
        axes[1].set_title('Uncertainty (σ)', fontsize=12, weight='bold')

        axes[1].scatter(X_den[:, d1], X_den[:, d2], s=point_sizes, c=point_colors,
                       cmap='Blues', vmin=0, vmax=1, alpha=0.6, ec='white', lw=1, zorder=5)

        if np.any(on_slice):
            axes[1].scatter(X_den[on_slice, d1], X_den[on_slice, d2], s=400, marker='*',
                           c='yellow', ec='black', lw=2, zorder=10, label='On slice')
            axes[1].legend()

        if d2 == 2: axes[1].axhline(0, color='green', ls='--', alpha=0.7, lw=2)
        if d1 == 2: axes[1].axvline(0, color='green', ls='--', alpha=0.7, lw=2)
        plt.colorbar(im2, ax=axes[1], label='Std Dev (σ)')

        plt.tight_layout()
        clean_name = slice_name.replace('#', 'num').replace('(', '').replace(')', '').replace('=', '').replace(':', '').replace('/', '_').replace(' ', '_')
        plt.savefig(f'ibeam_slice_{slice_idx:02d}_{clean_name}_{n1}_{n2}_B16mm.png', dpi=150, bbox_inches='tight')
        plt.show()

def optimize_acq_engineered(gp, acq_fn, acq_params, best_beam_params, n_cand=4000):
    """
    Acquisition optimization with engineering constraints:
    - b < best_b (thinner webs than current best)
    - ΔH > best_ΔH (closer to physics optimum)
    - r maximized for given b (always use max allowable radius)
    """
    b_best, dH_best = best_beam_params['b'], best_beam_params['dH']

    valid_cand, valid_norm = [], []
    attempts = 0
    max_attempts = n_cand * 20  # Allow more attempts due to tighter constraints

    print(f"    Constraints: b < {b_best:.2f}mm, ΔH > {dH_best:.2f}mm, r maximized")

    while len(valid_cand) < n_cand and attempts < max_attempts:
        attempts += 1

        # Sample b from constrained range
        b = np.random.uniform(BOUNDS_3D['b'][0], min(b_best, BOUNDS_3D['b'][1]))

        # Maximize radius for this b (with small random perturbation to avoid exact duplicates)
        r_max = min((B_FIXED - b) / 2.0, BOUNDS_3D['r'][1])
        r = np.random.uniform(max(0.95 * r_max, 0), r_max)  # 95-100% of max radius

        # Sample ΔH from constrained range (higher than best)
        dH = np.random.uniform(max(dH_best, BOUNDS_3D['delta_H'][0]), BOUNDS_3D['delta_H'][1])

        if check_3d(b, r, dH):
            cand = np.array([b, r, dH])
            valid_cand.append(cand)
            valid_norm.append(normalize(cand.reshape(1, -1))[0])

    if not valid_cand:
        raise ValueError(f"No valid candidates with b<{b_best:.2f}, ΔH>{dH_best:.2f}!")

    print(f"    Generated {len(valid_cand)} valid candidates from {attempts} attempts")

    valid_norm = np.array(valid_norm)
    acq_vals, mu_vals, sig_vals = acq_fn(gp, valid_norm, **acq_params)
    idx = np.argmax(acq_vals)
    return valid_cand[idx], acq_vals[idx], mu_vals[idx], sig_vals[idx]



def gen_engineered(gp, y_cent, y_mean, X_train_3d, y_train, best_beam_idx):
    """
    Generate EI recommendations with engineering constraints:
    - B fixed at 16mm
    - r maximized for given b (always use max allowable radius)
    - b and ΔH free to optimize
    """
    np.random.seed(100)
    print(f"\n{'='*70}\nENGINEERED EI: Maximize radius, free b and ΔH\n{'='*70}")

    print(f"B fixed at {B_FIXED}mm")
    print(f"Forcing r to 95-100% of maximum for each b")
    print(f"b and ΔH free to optimize")

    recs = []

    # Use original GP for all beams (non-sequential)
    for i in range(3):
        # Use EI with moderate exploration
        xi = ACQ_CONFIG['EI']['moderate']
        y_best = y_cent.max()

        # Generate candidates with maximized r
        valid_cand, valid_norm = [], []
        attempts = 0
        max_attempts = 4000 * 20

        print(f"  Beam {i+1}: Searching with r maximized, b and ΔH free")

        while len(valid_cand) < 4000 and attempts < max_attempts:
            attempts += 1

            # Sample b from full range
            b = np.random.uniform(*BOUNDS_3D['b'])

            # Maximize radius for this b
            r_max = min((B_FIXED - b) / 2.0, BOUNDS_3D['r'][1])
            r = np.random.uniform(max(0.95 * r_max, 0), r_max)  # 95-100% of max

            # Sample ΔH from full range
            dH = np.random.uniform(*BOUNDS_3D['delta_H'])

            if check_3d(b, r, dH):
                cand = np.array([b, r, dH])
                valid_cand.append(cand)
                valid_norm.append(normalize(cand.reshape(1, -1))[0])

        if not valid_cand:
            raise ValueError(f"No valid candidates found!")

        print(f"    Generated {len(valid_cand)} valid candidates from {attempts} attempts")

        # Run acquisition function
        valid_norm = np.array(valid_norm)
        acq_vals, mu_vals, sig_vals = ei_acq(gp, valid_norm, y_best, xi)
        idx = np.argmax(acq_vals)
        x = valid_cand[idx]
        mu, sig = mu_vals[idx], sig_vals[idx]

        x, H, h = enforce_3d(x)
        pred = np.exp(mu + y_mean)

        recs.append({
            'Beam': i+1, 'Method': 'EI_Eng',
            'Param': f'ξ={xi:.4f}', 'Type': 'MaxR',
            'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
            'Pred': pred, 'Var': sig
        })

    return pd.DataFrame(recs)


def gen_ucb_engineered(gp, y_mean, X_train_3d, y_train, best_beam_idx):
    """
    Generate UCB recommendations with engineering constraints:
    - B fixed at 16mm
    - r maximized for given b (always use max allowable radius)
    - b and ΔH free to optimize
    """
    np.random.seed(100)
    print(f"\n{'='*70}\nENGINEERED UCB: Maximize radius, free b and ΔH\n{'='*70}")

    print(f"B fixed at {B_FIXED}mm")
    print(f"Forcing r to 95-100% of maximum for each b")
    print(f"b and ΔH free to optimize")

    recs = []
    kappa = ACQ_CONFIG['UCB']['moderate']

    # Use original GP for all beams (non-sequential)
    for i in range(3):
        # Generate candidates with maximized r
        valid_cand, valid_norm = [], []
        attempts = 0
        max_attempts = 4000 * 20

        print(f"  Beam {i+1}: Searching with r maximized, b and ΔH free")

        while len(valid_cand) < 4000 and attempts < max_attempts:
            attempts += 1

            # Sample b from full range
            b = np.random.uniform(*BOUNDS_3D['b'])

            # Maximize radius for this b
            r_max = min((B_FIXED - b) / 2.0, BOUNDS_3D['r'][1])
            r = np.random.uniform(max(0.95 * r_max, 0), r_max)  # 95-100% of max

            # Sample ΔH from full range
            dH = np.random.uniform(*BOUNDS_3D['delta_H'])

            if check_3d(b, r, dH):
                cand = np.array([b, r, dH])
                valid_cand.append(cand)
                valid_norm.append(normalize(cand.reshape(1, -1))[0])

        if not valid_cand:
            raise ValueError(f"No valid candidates found!")

        print(f"    Generated {len(valid_cand)} valid candidates from {attempts} attempts")

        # Run acquisition function
        valid_norm = np.array(valid_norm)
        acq_vals, mu_vals, sig_vals = ucb_acq(gp, valid_norm, kappa)
        idx = np.argmax(acq_vals)
        x = valid_cand[idx]
        mu, sig = mu_vals[idx], sig_vals[idx]

        x, H, h = enforce_3d(x)
        pred = np.exp(mu + y_mean)

        recs.append({
            'Beam': i+1, 'Method': 'UCB_Eng',
            'Param': f'κ={kappa:.4f}', 'Type': 'MaxR',
            'H': H, 'B': B_FIXED, 'b': x[0], 'r': x[1], 'h': h, 'dH': x[2],
            'Pred': pred, 'Var': sig
        })

    return pd.DataFrame(recs)
# ==========================================
# DATA LOADING
# ==========================================
def load_initial_data_from_github():
    """Load tested beams from GitHub CSV"""
    import urllib.request
    import io

    DATA_URL = "https://raw.githubusercontent.com/andrewvoss8-boop/core-me-data-science-activities-public/main/data/I_beam_data.csv"

    print(f"📥 Downloading data from GitHub...")
    try:
        with urllib.request.urlopen(DATA_URL) as response:
            csv_data = response.read().decode('utf-8')

        df = pd.read_csv(io.StringIO(csv_data))
        print(f"✓ Successfully loaded {len(df)} beams from GitHub")

        param_names_4d = ['H_web_height', 'B_flange_width', 'b_web_thick', 'r_fillet']
        required_cols = param_names_4d + ['Str/w (N/g)']

        missing_cols = [col for col in param_names_4d if col not in df.columns]
        if missing_cols:
            print(f"⚠️  Missing columns: {missing_cols}")
            return None

        df_clean = df[required_cols].dropna()
        print(f"✓ After removing NaN: {len(df_clean)} valid beams")

        def meets_constraints(row):
            H = row['H_web_height']
            B = row['B_flange_width']
            b = row['b_web_thick']
            r = row['r_fillet']
            h = (TOTAL_HEIGHT - H) / 2.0

            if B < MIN_FLANGE_WIDTH or b < MIN_WEB_THICKNESS or b > MAX_WEB_RATIO * B:
                return False
            if r < 0 or r > (B - b) / 2.0:
                return False
            if h < 0 or h > 6.5:
                return False
            return True

        df_constrained = df_clean[df_clean.apply(meets_constraints, axis=1)].copy()
        print(f"✓ After geometric filtering: {len(df_constrained)} beams")

        if len(df_constrained) > 0:
            str_w = df_constrained['Str/w (N/g)']
            print(f"   Str/w range: [{str_w.min():.2f}, {str_w.max():.2f}] N/g")

        return df_constrained

    except Exception as e:
        print(f"⚠️  Error loading from GitHub: {e}")
        return None


# ==========================================
# MAIN
# ==========================================
if __name__ == "__main__":
    print("="*70)
    print("PHYSICS-INFORMED I-BEAM OPTIMIZER")
    print("="*70)
    print(f"Explore/Exploit Dial: {EXPLORE_EXPLOIT_DIAL:.2f}")
    print(f"  EI ξ_moderate: {ACQ_CONFIG['EI']['moderate']:.4f}")
    print(f"  UCB κ_moderate: {ACQ_CONFIG['UCB']['moderate']:.4f}")

    # Load data from GitHub
    df = load_initial_data_from_github()

    if df is None or len(df) == 0:
        print("⚠️ Failed to load data. Exiting.")
        exit(1)

    # Extract arrays using the exact column names from GitHub data
    X_4d = df[['H_web_height', 'B_flange_width', 'b_web_thick', 'r_fillet']].values
    y = df['Str/w (N/g)'].values
    print(f"✓ Loaded {len(X_4d)} beams, Str/w range: [{y.min():.2f}, {y.max():.2f}]")

    # Transform & train
    print("\nTransforming to 3D...")
    X_3d = transform_4d_to_3d(X_4d)
    print(f"✓ Transformed. Sample: b={X_3d[0,0]:.2f}, r={X_3d[0,1]:.2f}, dH={X_3d[0,2]:.2f}")

    print("\nTraining GP...")
    gp, X_norm, y_cent, y_mean = train_gp(X_3d, y)
    scales = gp.kernel_.k2.length_scale
    print(f"✓ GP trained. ARD scales: b={scales[0]:.3f}, r={scales[1]:.3f}, dH={scales[2]:.3f}")


# Find best beam
    best_idx = np.argmax(y)
    print(f"\nCurrent best beam (index {best_idx}):")
    print(f"  b={X_3d[best_idx, 0]:.2f}mm, r={X_3d[best_idx, 1]:.2f}mm, " +
      f"ΔH={X_3d[best_idx, 2]:.2f}mm → Str/w={y[best_idx]:.2f} N/g")

# Generate standard recommendations
    rec_mixed = gen_mixed(gp, y_cent, y_mean, X_3d, y) if GENERATE_MIXED else None
    rec_ucb = gen_ucb(gp, y_mean, X_3d, y, y_cent) if GENERATE_UCB else None
    rec_ei = gen_ei(gp, y_cent, y_mean, X_3d, y) if GENERATE_EI else None

# Generate engineered recommendations
    rec_engineered_ei = gen_engineered(gp, y_cent, y_mean, X_3d, y, best_idx)
    rec_engineered_ucb = gen_ucb_engineered(gp, y_mean, X_3d, y, best_idx)
# Generate engineered recommendations
# Add to printing loop:
    for name, rec in [('MIXED', rec_mixed),
                  ('UCB', rec_ucb),
                  ('EI', rec_ei),
                  ('ENGINEERED_EI', rec_engineered_ei),
                  ('ENGINEERED_UCB', rec_engineered_ucb)]:
        if rec is not None:
            print(f"\n{'='*70}\n{name} RECOMMENDATIONS\n{'='*70}")
            print(rec.to_string(index=False))
            rec.to_csv(f'recommendations_{name.lower()}.csv', index=False)
            print(f"✓ Saved to recommendations_{name.lower()}.csv")

    # MARGINAL PLOTS (TRUE AVERAGES)
    print("\n" + "="*70)
    print("GENERATING MARGINAL PLOTS (AVERAGED ACROSS DESIGN SPACE)")
    print("="*70)
    plot_1d_marginal(gp, X_norm, y, y_mean, [rec_mixed, rec_ucb, rec_ei], n_samples=200)
    plot_2d_marginal(gp, X_norm, y, y_mean, [rec_mixed, rec_ucb, rec_ei], n_samples=100)

    # GLOBAL SLICE PLOTS
    print("\n" + "="*70)
    print("GENERATING GLOBAL SLICE PLOTS")
    print("="*70)
    plot_1d_global(gp, X_norm, y, y_mean, [rec_mixed, rec_ucb, rec_ei])
    plot_2d_global(gp, X_norm, y, y_mean, [rec_mixed, rec_ucb, rec_ei])

    # SLICE PLOTS
    print("\n" + "="*70)
    print("GENERATING SLICE-SPECIFIC PLOTS")
    print("="*70)

    slices = []

    # Top 3 strongest beams
    top3_indices = np.argsort(y)[-3:][::-1]
    for i, idx in enumerate(top3_indices):
        slice_name = f"Top #{i+1} (Str/w={y[idx]:.2f} N/g)"
        slices.append((X_3d[idx], slice_name))

# Add all recommendations
    all_recs = []
    if rec_mixed is not None:
        for i, row in rec_mixed.iterrows():
            all_recs.append((np.array([row['b'], row['r'], row['dH']]),
                           f"Mixed Beam {row['Beam']} ({row['Method']})"))
    if rec_ucb is not None:
        for i, row in rec_ucb.iterrows():
            all_recs.append((np.array([row['b'], row['r'], row['dH']]),
                          f"UCB Beam {row['Beam']}"))
    if rec_ei is not None:
        for i, row in rec_ei.iterrows():
            all_recs.append((np.array([row['b'], row['r'], row['dH']]),
                          f"EI Beam {row['Beam']}"))
    if rec_engineered_ei is not None:
        for i, row in rec_engineered_ei.iterrows():
            all_recs.append((np.array([row['b'], row['r'], row['dH']]),
                          f"Engineered EI Beam {row['Beam']}"))
    if rec_engineered_ucb is not None:
        for i, row in rec_engineered_ucb.iterrows():
            all_recs.append((np.array([row['b'], row['r'], row['dH']]),
                          f"Engineered UCB Beam {row['Beam']}"))

    slices.extend(all_recs)

    # Normalize slice points
    for idx, (slice_point_3d, slice_name) in enumerate(slices):
        slice_point_norm = normalize(slice_point_3d.reshape(1, -1))[0]
        print(f"\nProcessing slice {idx+1}/{len(slices)}: {slice_name}")
        plot_slice_1d(gp, X_norm, y, y_mean, slice_point_norm, slice_name, idx+1)
        plot_slice_2d(gp, X_norm, y, y_mean, slice_point_norm, slice_name, idx+1)

    print("\n" + "="*70)
    print("✓ COMPLETE! All plots generated.")
    print("="*70)
    print(f"Generated:")
    print(f"  - 1 global 1D plot")
    print(f"  - 3 global 2D contour pairs")
    print(f"  - {len(slices)} slice-specific 1D plots")
    print(f"  - {len(slices)*3} slice-specific 2D contour pairs")
    print(f"  Total: {1 + 3 + len(slices)*4} figures")
    print("="*70)
```
